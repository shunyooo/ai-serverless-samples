# SD Sample

### Call via Modal SDK

```python
from PIL.Image import Image as PILImage
import modal

f = modal.Cls.from_name("example-text-to-image", "SD")().run
prompt = "photo of a huge white cat sitting on a cloud in the sky"
images: list[PILImage] = f.remote(prompt=prompt)

for image in images:
    image.resize((512, 512)).show()
```
